# Set-up

```python
import os
import GEOparse
```

```python
# Choose the current dataset we are working with
dataset_name = "AI-ATAC"
geo_id = "GSE100738"
```

```python
# Set-up directories
base_dir = "/cellar/users/aklie/data/datasets"
cwd = os.path.join(base_dir, dataset_name, "bin", "data_acquisition")
analysis_dir = os.path.join(base_dir, dataset_name, "analysis", "10Nov23", "geo")
base_dir, cwd, analysis_dir
```

```python
# Make the directory if it doesn't exist
if not os.path.exists(analysis_dir):
    os.makedirs(analysis_dir)
```

```python
# Parse GEO
gse = GEOparse.get_GEO(geo=geo_id, destdir=cwd)
```

```python
# Check metadata
gse.metadata
```

```python
# for every file in the gse.metadata "supplementary_file" key, download it to the annotation directory
for file in gse.metadata["supplementary_file"]:
    cmd = f"wget -P {analysis_dir} {file}"
    print(cmd)
    os.system(cmd)
```
