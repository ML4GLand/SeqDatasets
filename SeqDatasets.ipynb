### `lee22`

```python
# Get the paths of all the datasets that end in h5, exclude "CAGI_onehot.h5", "Basset_data.h5" and "DeepSTARR_data.h5"
dataset_dir = "/cellar/users/aklie/projects/ML4GLand/data/pubs/lee22"
dataset_paths = glob.glob(os.path.join(dataset_dir, "*.h5"))
dataset_paths = [path for path in dataset_paths if path.split("/")[-1] not in ["CAGI_onehot.h5", "Basset_data.h5", "DeepSTARR_data.h5"]]
dataset_paths 
```

```python
def load_train_set_from_pkoo_h5(
    h5_file
):
    dataset = h5py.File(h5_file, 'r')
    x_train = np.array(dataset['x_train']).astype(np.float32)
    y_train = np.array(dataset['y_train']).astype(np.float32)
    x_train = np.squeeze(x_train).transpose([0,2,1])
    y_train = np.squeeze(y_train)
    dataset.close()
    return x_train, y_train
```

```python
for dataset in dataset_paths:
    dataset_name = dataset.split("/")[-1].split(".")[0]
    x_train, y_train = load_train_set_from_pkoo_h5(dataset)
    sdata = sd.SeqData(
        ohe_seqs=x_train,
        names=np.array([f"seq_{i}" for i in range(x_train.shape[0])]),
        seqs_annot=pd.DataFrame(y_train, columns=["target"]),
    )
    sdata.write_h5sd(os.path.join(seqdatasets_dir, workflow, f"lee22_{dataset_name}_train.h5sd"))
    with open(os.path.join(seqdatasets_dir, "README.md"), "a") as f:
        f.write(f"## Lee et al. 2022 - {dataset_name} train set\n")
        f.write(f"This dataset contains the {dataset_name} train set from Lee et al. 2022.\n")
        f.write(f"The dataset was preprocessed as described in the EUGENe paper.\n")
        f.write(f"The dataset contains {sdata.n_obs} samples.\n")
        f.write(f"Each sequence is {len(sdata.ohe_seqs[0])} nucleotides long.\n")
        f.write("The data is saved in the h5sd format.\n")
```

```python
def guess_key(f):
    """Guess key name from the file.keys()."""
    for key in f.keys():
        if key.endswith("xdata"):
            return key[:-5]
    return ""
    
def load_from_mat_or_npy(filename, key=None, load_data=True, load_labels = True):
    """Load labels and/or data from the .mat or .npy file."""
    data = None
    labels = None
    if filename.endswith(".mat"):
        try:
            f = h5py.File(filename, 'r')
        except OSError:
            # probably an old MATLAB file. try opening with scipy
            from scipy.io import loadmat
            f = loadmat(filename)
        key = key or guess_key(f)
        if load_data:
            data = f[f"{key}xdata"]
        if load_labels:
            labels = f[f"{key}data"]
            print(labels.shape)
    else:
        if load_data:
            data = np.load(filename)
        if load_labels:
            labels = np.load(filename)
    return data, labels
```

```python
def load_train_set_from_pkoo_h5(
    h5_file,
    upper_case=False
):
    if upper_case:
        X_key = "X_train"
        y_key = "Y_train"
    else:
        X_key = "x_train"
        y_key = "y_train"
    dataset = h5py.File(h5_file, 'r')
    x_train = np.array(dataset[X_key]).astype(np.float32)
    y_train = np.array(dataset[y_key]).astype(np.float32)
    x_train = np.squeeze(x_train).transpose([0,2,1])
    y_train = np.squeeze(y_train)
    dataset.close()
    return x_train, y_train
```

```python
y_train.shape
```

```python
dataset_dir = "/cellar/users/aklie/projects/ML4GLand/data/pubs/lee22"
dataset_path = os.path.join(dataset_dir, "Basset_data.h5")
x_train, y_train = load_train_set_from_pkoo_h5(dataset_path, upper_case=True)
sdata = sd.SeqData(
    ohe_seqs=x_train,
    names=np.array([f"seq_{i}" for i in range(x_train.shape[0])]),
    seqs_annot=pd.DataFrame(y_train, columns=[f"target_{i}" for i in range(y_train.shape[1])]),
)
sdata.write_h5sd(os.path.join(seqdatasets_dir, workflow, f"lee22_Basset_train.h5sd"))
```
